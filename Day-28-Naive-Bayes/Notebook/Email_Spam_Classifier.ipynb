```python
# Import pandas to create and manage tabular data
import pandas as pd

# Split the dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Convert email text into numerical word counts
from sklearn.feature_extraction.text import CountVectorizer

# Import the Multinomial Naive Bayes classification algorithm
from sklearn.naive_bayes import MultinomialNB

# Import evaluation metrics to measure model performance
from sklearn.metrics import (
    accuracy_score,       # Calculate the percentage of correct predictions
    confusion_matrix,     # Count correct and incorrect predictions
    classification_report # Display precision, recall, and F1-score
)
```

```python
# Create a dictionary containing sample emails and their labels
data = {
    "email": [
        "win a free prize now",
        "claim your free gift today",
        "meeting scheduled for tomorrow",
        "please send the project report"
    ],

    # Label each email as spam or not spam
    "label": [
        "spam",
        "spam",
        "not_spam",
        "not_spam"
    ]
}

# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the first five rows to inspect the dataset
df.head()
```

```python
# Display the number of rows and columns
print("Dataset shape:", df.shape)

# Display the column names and data types
print("\nDataset information:")
df.info()

# Count the number of emails in each class
print("\nEmail class distribution:")
print(df["label"].value_counts())

# Check for missing values in each column
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Store email text as the input feature
X = df["email"]

# Store spam labels as the target variable
y = df["label"]

# Display the first five input emails
print("Input emails:")
print(X.head())

# Display the first five target labels
print("\nTarget labels:")
print(y.head())
```

```python
# Split the dataset into 75% training and 25% testing
# random_state=42 makes the split reproducible
# Remove stratify temporarily to avoid class-count split errors
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

# Display the sizes of the resulting datasets
print("Training emails:", len(X_train))
print("Testing emails:", len(X_test))

# Check the class distribution in both sets
print("\nTraining label counts:")
print(y_train.value_counts())

print("\nTesting label counts:")
print(y_test.value_counts())
```

```python
# Create a vectorizer that counts words in each email
vectorizer = CountVectorizer()

# Learn the vocabulary from training emails and convert them into word counts
X_train_counts = vectorizer.fit_transform(X_train)

# Convert test emails using the vocabulary learned from training data
X_test_counts = vectorizer.transform(X_test)

# Display the shape of the numerical training matrix
print("Training matrix shape:", X_train_counts.shape)

# Display the shape of the numerical testing matrix
print("Testing matrix shape:", X_test_counts.shape)
```

```python
# Create the Multinomial Naive Bayes classifier
model = MultinomialNB()

# Train the model using word counts and their corresponding labels
model.fit(X_train_counts, y_train)

# Display a success message after training
print("Naive Bayes model trained successfully!")
```

```python
# Predict the labels of emails in the testing dataset
y_pred = model.predict(X_test_counts)

# Create a DataFrame to compare actual and predicted labels
results = pd.DataFrame({
    "Email": X_test.values,
    "Actual Label": y_test.values,
    "Predicted Label": y_pred
})

# Display the prediction results
display(results)
```

```python
# Calculate the percentage of correct predictions
accuracy = accuracy_score(y_test, y_pred)

# Display the model accuracy
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
# Compare actual labels with predicted labels
# Keep the label order consistent for easier interpretation
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["not_spam", "spam"]
)

# Display the confusion matrix
print("Confusion Matrix:")
print(cm)

# Explain the matrix layout
print("\nRows represent actual labels.")
print("Columns represent predicted labels.")
print("Label order: not_spam, spam")
```
