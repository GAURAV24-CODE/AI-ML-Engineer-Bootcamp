```python
#task 1-Create a NumPy Array from 1 to 50
# Import NumPy
import numpy as np

# Create an array from 1 to 50
arr = np.arange(1, 51)

# Display the array
print(arr)
```

```python
#task 2 -Find the Sum and Average
# Calculate the sum
print("Sum:", np.sum(arr))

# Calculate the average
print("Average:", np.mean(arr))
```

```python
#task 3 -Reshape into a 5 × 10 Matrix
# Reshape the array into 5 rows and 10 columns
matrix = arr.reshape(5, 10)

# Display the matrix
print(matrix)
```

```python
#task  4 – Generate 15 Random Numbers
# Generate 15 random integers between 1 and 100
random_numbers = np.random.randint(1, 101, 15)

# Display the random numbers
print(random_numbers)
```

```python
#task 5 – Sort the Array
# Sort the random numbers
sorted_array = np.sort(random_numbers)

# Display the sorted array
print(sorted_array)
```

```python
#task 6 – Find Unique Values
# Create an array with duplicate values
numbers = np.array([1, 2, 2, 3, 4, 4, 5, 6, 6, 7])

# Find unique values
unique_values = np.unique(numbers)

# Display unique values
print(unique_values)
```

```python
#task 7 – Stack Two Arrays
# Create two arrays
a = np.array([1, 2, 3])

b = np.array([4, 5, 6])

# Stack the arrays vertically
stacked = np.vstack((a, b))

# Display the stacked array
print(stacked)
```

```python

#task 8 – Split an Array into 5 Parts
# Create an array
arr = np.arange(1, 26)

# Split the array into 5 equal parts
split_array = np.array_split(arr, 5)

# Display the split arrays
print(split_array)
```

```python
#task 9 – Calculate Standard Deviation
# Create an array
marks = np.array([80, 75, 90, 85, 95])

# Calculate standard deviation
std = np.std(marks)

# Display the result
print("Standard Deviation:", std)
```

```python
#task 10 – Create an Identity Matrix of Size 5
# Create a 5 x 5 identity matrix
identity_matrix = np.eye(5)

# Display the identity matrix
print(identity_matrix)
```
