```python
import numpy as np

print("NumPy Version:", np.__version__)
```

```python
# Create a 1D NumPy array

arr = np.array([10, 20, 30, 40, 50])

print(arr)
print(type(arr))
```

```python
# Create a 1D NumPy array

arr = np.array([10, 20, 30, 40, 50])

print(arr)
print(type(arr))
```

```python
# Create a 2D array

arr_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr_2d)
```

```python
# Create zeros, ones and full arrays

print(np.zeros(5))
print(np.ones(5))
print(np.full((2, 3), 7))
```

```python
# Create sequences using arange and linspace

print("arange:", np.arange(1, 11, 2))
print("linspace:", np.linspace(0, 10, 5))
```

```python
# Check array properties

print("Dimensions:", arr_2d.ndim)
print("Shape:", arr_2d.shape)
print("Size:", arr_2d.size)
print("Data Type:", arr_2d.dtype)
```

```python
# Access array elements using indexing

print("First:", arr[0])
print("Third:", arr[2])
print("Last:", arr[-1])
```

```python
# Extract part of an array
print(arr)
print(arr[1:4])
print(arr[:3])
print(arr[::2])
```

```python
# Perform vectorized arithmetic operations

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Filter values using conditions

marks = np.array([45, 78, 92, 34, 67, 88])

print("Above 70:", marks[marks > 70])
print("Passed:", marks[marks >= 50])
```

```python
# Apply an operation to every array element

print("Original:", marks)
print("After adding 5:", marks + 5)
```

```python
# Calculate basic statistics
print("Original:", marks)
print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard Deviation:", np.std(marks))
```

```python
# Find positions of maximum and minimum values

print("Maximum index:", np.argmax(marks))
print("Minimum index:", np.argmin(marks))
```

```python
# Change the shape of an array

numbers = np.arange(1, 13)

matrix = numbers.reshape(3, 4)

print(matrix)
```

```python
# Flatten and transpose a matrix

print("Flatten:")
print(matrix.flatten())

print("\nTranspose:")
print(matrix.T)
```

```python
# Perform calculations row-wise and column-wise

print("Column Sum:", np.sum(matrix, axis=0))
print("Row Sum:", np.sum(matrix, axis=1))
print("Row Mean:", np.mean(matrix, axis=1))
```
