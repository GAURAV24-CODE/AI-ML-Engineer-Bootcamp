# 📌 Types of Function Arguments

Python supports different types of arguments:

- Default Arguments
- Keyword Arguments
- Variable-Length Arguments (*args)
- Keyword Variable-Length Arguments (**kwargs)

These make functions flexible and reusable.

```python
# Default Argument

def country(name="India"):
    print("Country:", name)

country_name = input("Enter country (Press Enter for India): ")

if country_name == "":
    country()
else:
    country(country_name)

print()

# Keyword Arguments

name = input("Enter student name: ")
age = int(input("Enter student age: "))

def student(name, age):
    print(f"Name : {name}")
    print(f"Age  : {age}")

student(age=age, name=name)

print()

# *args

numbers = list(map(int, input("Enter numbers separated by space: ").split()))

def total(*nums):
    print("Total =", sum(nums))

total(*numbers)

print()

# **kwargs

name = input("Enter Name: ")
course = input("Enter Course: ")
city = input("Enter City: ")

def details(**info):
    for key, value in info.items():
        print(f"{key}: {value}")

details(Name=name, Course=course, City=city)
```

# 📌 Lambda Function & Recursion

### Lambda Function
A lambda function is a small anonymous function.

### Recursion
A recursive function is a function that calls itself until a stopping condition is reached.

```python
# Lambda Function

num = int(input("Enter a number to find its square: "))

square = lambda x: x ** 2

print("Square =", square(num))

print()

# Recursion

n = int(input("Enter a number to find factorial: "))

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print("Factorial =", factorial(n))
```

# 📌 Python Modules

Modules help organize code and provide useful built-in functions.

Common modules:
- math
- random
- datetime

```python
import math
import random
from datetime import datetime

num = float(input("Enter a number to find square root: "))

print("Square Root =", math.sqrt(num))
print("Value of π =", math.pi)

start = int(input("Enter random range start: "))
end = int(input("Enter random range end: "))

print("Random Number =", random.randint(start, end))

print("Current Date & Time =", datetime.now())
```
