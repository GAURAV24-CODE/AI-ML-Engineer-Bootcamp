Data Cleaning means finding and fixing problems in raw data.

Topics:
- Missing Values
- Duplicates
- Text Cleaning
- Data Types
- Filtering
- Sorting
- GroupBy
- Outliers

```python
# Import Pandas 
import pandas as pd

# Import NumPy
import numpy as np

```

```python
# Create a sample dataset 
data = {
    "Name": [" Gaurav", "Rahul ", "Priya", "Amit", "Gaurav"],
    "Age": [21, np.nan, 23, 25, 21],
    "Gender": ["M", "Male", "F", "Female", "M"],
    "Department": ["IT", "it", "HR", "Finance", "IT"],
    "Salary": [50000, 60000, np.nan, 55000, 50000]
}


df = pd.DataFrame(data)


df
```

```python
# Display the first 5 rows
print(df.head())

# Check number of rows and columns
print("\nShape:", df.shape)
```

```python
print("\nColumns:", df.columns)
print("----------------------------------------------")

# Get information about columns, data types and missing value
df.info()
```

```python
# Get numerical statistics such as mean, min and max
print("\nStatistics:")
print(df.describe())
```

```python
# Check every cell for a missing value
print(df.isnull())

# Count missing values in each column
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Fill missing Age values with the median Age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary values with the median Salary
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Display cleaned data
df
```

```python
# Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Display the result
df
```

```python
# Remove spaces before and after names
df["Name"] = df["Name"].str.strip()
```

```python
df
```

```python
# Convert M/F into full gender names
df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "F": "Female"
})
df
```

```python
# Show unique gender values
print("Gender:")
print(df["Gender"].unique())
```

```python
# Show unique department values
print("\nDepartment:")
print(df["Department"].unique())
```

```python
# Count each gender
print("\nGender Counts:")
print(df["Gender"].value_counts())
```

```python
# Convert salary from rupees to lakhs
df["Salary_Lakh"] = df["Salary"] / 100000

# Display the DataFrame
df
```

```python
# Select employees whose salary is greater than 50,000
high_salary = df[df["Salary"] > 50000]

high_salary
```

```python
# Sort employees from highest to lowest salary
df.sort_values("Salary", ascending=False)
```

```python
# Find the first quartile
Q1 = df["Salary"].quantile(0.25)

# Find the third quartile
Q3 = df["Salary"].quantile(0.75)

# Calculate the Interquartile Range
IQR = Q3 - Q1

# Calculate acceptable lower and upper limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
# Group employees by department
# Then calculate the average salary
df.groupby("Department")["Salary"].mean()
```

```python
# Check number of rows and columns
print("Shape:", df.shape)

# Check remaining missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check remaining duplicates
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Check data types
print("\nData Types:")
print(df.dtypes)

# Display final cleaned dataset
display(df)
```
