```python
# Import required libraries
import pandas as pd
import numpy as np

# Create the dataset
data = {
    "Name": [" Gaurav", "Rahul ", "Priya", "Amit", "Gaurav", "Sneha"],
    "Age": [21, np.nan, 23, 25, 21, np.nan],
    "Gender": ["M", "Male", "F", "Female", "M", "F"],
    "Department": ["IT", "it", "HR", "Finance", "IT", "hr"],
    "Salary": [50000, 60000, np.nan, 55000, 50000, 70000]
}

df = pd.DataFrame(data)

df
```

```python
# Task 1: Display dataset information
print(df.head())
print("-----------------------------")
print("shapes:",df.shape)

print("-----------------------------")
print(df.dtypes)
```

```python
# Task 2: Find missing values
print(df.isnull().sum())
```

```python
# Task 3: Fill missing Age and Salary
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print(df)
```

```python
# Task 4: Remove duplicate rows
df = df.drop_duplicates()
print(df)
```

```python
# Task 5: Clean Name, Gender and Department
df["Name"] = df["Name"].str.strip()
df["Gender"] = df["Gender"].replace({"M":"Male", "F":"Female"})
df["Department"] = df["Department"].str.strip().str.title()
print(df)
```

```python
# Task 6: Convert Age and Salary to integer
df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].astype(int)
print(df.dtypes)
```

```python
# Task 7: Filter and sort salary
print(df[df["Salary"] > 50000])
print("-----------------------------")
print(df.sort_values("Salary", ascending=False))
```

```python
# Task 8: GroupBy department
print(df["Department"].value_counts())
print("-----------------------------")
print(df.groupby("Department")["Salary"].mean())
```

```python
# Task 9: Find salary outliers using IQR
Q1 = df["Salary"].quantile(.25)
Q3 = df["Salary"].quantile(.75)
IQR = Q3 - Q1

outliers = df[(df["Salary"] < Q1-1.5*IQR) | (df["Salary"] > Q3+1.5*IQR)]
print(outliers)
```

```python
# Task 10: Final data validation
print("Missing:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
print("Shape:", df.shape)
print(df)
```
