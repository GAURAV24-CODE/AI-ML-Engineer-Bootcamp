```python
# ----------------------------------------
# Creating a File
# ----------------------------------------
# "w" mode creates a new file.
# If the file already exists,
# it overwrites the previous content.

file = open("sample.txt", "w")

file.write("Hello Python!")

file.close()

print("File created successfully.")
```

```python
file = open("sample.txt", "r")

print(file.read())

file.close()
```

```python
file = open("sample.txt", "w")

file.write("Learning File Handling")

file.close()

print("Data written successfully.")
```

```python
file = open("sample.txt", "a")

file.write("\nWelcome to AI/ML Bootcamp")

file.close()

print("Data appended successfully.")
```

```python
# ----------------------------------------
# Reading Updated File
# ----------------------------------------

file = open("sample.txt", "r")

print(file.read())

file.close()
```

```python
with open("sample.txt", "r") as file:

    print(file.read())
```

```python
with open("student.txt", "w") as file:

    file.write("Rahul\n")
    file.write("Priya\n")
    file.write("Amit\n")
    file.write("Neha\n")

print("Student file created.")
```

```python
with open("student.txt", "a") as file:

    file.write("Rohit\n")

print("New student added.")
```

```python
name = input("Enter your name: ")

with open("user.txt", "w") as file:

    file.write(name)

print("User name saved successfully.")
```

```python
with open("user.txt", "r") as file:

    print(file.read())
```

```python
sentence = input("Enter a sentence: ")

with open("sentence.txt", "w") as file:

    file.write(sentence)

with open("sentence.txt", "r") as file:

    words = file.read().split()

print("Total Words:", len(words))
print(words)
```
