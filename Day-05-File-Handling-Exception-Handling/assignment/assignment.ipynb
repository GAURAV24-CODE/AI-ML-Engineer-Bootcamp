```python
# Question 1 Create a file named "intro.txt" and write two lines.

with open("intro.txt", "w") as file:

    file.write("Hello Python\n")
    file.write("Welcome to File Handling")

print("File created successfully.")
```

```python
# Question 2 Read and display the contents of "intro.txt"

with open("intro.txt", "r") as file:

    print(file.read())
```

```python
# Question 3 Append the text:

with open("intro.txt", "a") as file:

    file.write("\nLearning Python is fun!")

print("Text appended successfully.")
```

```python
# Question 6 Create a file named "numbers.txt" and write numbers from 1 to 10.

with open("numbers.txt", "w") as file:

    for i in range(1, 11):

        file.write(str(i) + "\n")

print("Numbers written successfully.")
```

```python
# Question 7 Read all numbers from "numbers.txt".

with open("numbers.txt", "r") as file:

    print(file.read())
```

```python
# Question 9 Write five student names into "students.txt".

students = [
    "Gaurav",
    "Rahul",
    "Sneha",
    "Priya",
    "Amit"
]

with open("students.txt", "w") as file:

    for student in students:

        file.write(student + "\n")

print("Student names saved.")
```

```python
# Question 8 Handle the error if a file does not exist.

try:

    with open("data.txt", "r") as file:

        print(file.read())

except FileNotFoundError:

    print("File not found.")
```
