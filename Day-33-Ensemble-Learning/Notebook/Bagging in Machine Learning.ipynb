```python
# Topic: Bagging Ensemble Learning

import pandas as pd

# Split data into training and testing sets
from sklearn.model_selection import train_test_split

# Bagging ensemble classifier
from sklearn.ensemble import BaggingClassifier

# Decision Tree used as the base model
from sklearn.tree import DecisionTreeClassifier

# Evaluate model performance
from sklearn.metrics import accuracy_score
```

```python
# Create a small dataset for predicting fruit type
data = {
    "Weight": [120, 130, 140, 150, 160, 170, 180, 190, 125, 175],
    "Sweetness": [7, 6, 8, 5, 4, 3, 2, 3, 8, 2],
    "Fruit": [
        "Apple", "Apple", "Apple", "Apple", "Orange",
        "Orange", "Orange", "Orange", "Apple", "Orange"
    ]
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

```python
# X contains the input features
X = df[["Weight", "Sweetness"]]

# y contains the value we want to predict
y = df["Fruit"]

# Display the features and target
print("Features (X):")
print(X)

print("\nTarget (y):")
print(y)
```

```python
# Use 80% of rows for training and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the number of rows in each set
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
# Create one Decision Tree as our baseline model
tree_model = DecisionTreeClassifier(random_state=42)

# Train the tree on the training data
tree_model.fit(X_train, y_train)

# Predict fruit types for the test data
tree_predictions = tree_model.predict(X_test)

# Display the predictions
print("Decision Tree predictions:", tree_predictions)

# Evaluate the baseline accuracy
print(
    "Decision Tree accuracy:",
    round(accuracy_score(y_test, tree_predictions), 2)
)
```

```python
# Create a Bagging classifier using Decision Trees
bagging_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=10,       # Train 10 separate trees
    max_samples=1.0,       # Sample the full training-set size per tree
    bootstrap=True,         # Sample rows with replacement
    random_state=42,       # Make results reproducible
    n_jobs=-1               # Use available CPU cores
)

# Train the Bagging ensemble
bagging_model.fit(X_train, y_train)

print("Bagging model training completed!")
```

```python
# Predict fruit types using the ensemble of trees
bagging_predictions = bagging_model.predict(X_test)

# Display actual and predicted fruit labels
print("Actual values:", y_test.to_list())
print("Bagging predictions:", bagging_predictions)
```

```python
# Calculate the Bagging model's test accuracy
bagging_accuracy = accuracy_score(y_test, bagging_predictions)

# Display the result
print("Bagging accuracy:", round(bagging_accuracy, 2))
print("Bagging accuracy (%):", round(bagging_accuracy * 100, 2), "%")
```

```python
# Compare the single Decision Tree with Bagging
tree_accuracy = accuracy_score(y_test, tree_predictions)

results = pd.DataFrame({
    "Model": ["Decision Tree", "Bagging"],
    "Accuracy": [tree_accuracy, bagging_accuracy]
})

# Display the comparison table
print(results)
```

```python
# Predict the fruit type for a new example
# Weight = 135, Sweetness = 7
new_fruit = pd.DataFrame({
    "Weight": [135],
    "Sweetness": [7]
})

# Use the trained Bagging model
prediction = bagging_model.predict(new_fruit)

# Display the predicted fruit
print("Predicted fruit:", prediction[0])
```
