```python
# Create a copy so raw data remains unchanged
df_clean = df.copy()
print("Clean copy created.")
```

```python
# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()
print("Rows after removing duplicates:", len(df_clean))
```

```python
# Standardize region names

df_clean["Region"] = (
    df_clean["Region"]
    .str.strip()
    .str.title()
)

df_clean["Region"].value_counts()
```

```python
# Convert OrderDate to datetime

df_clean["OrderDate"] = pd.to_datetime(
    df_clean["OrderDate"],
    errors="coerce"
)

print(df_clean["OrderDate"].dtype)
```

```python
# Check invalid dates

print(
    "Invalid dates:",
    df_clean["OrderDate"].isna().sum()
)
```

```python
# Check invalid dates

print(
    "Invalid dates:",
    df_clean["OrderDate"].isna().sum()
)
```

```python
# Create Revenue

df_clean["Revenue"] = (
    df_clean["Quantity"]
    * df_clean["UnitPrice"]
)

df_clean.head()
```

```python
# Create date features

df_clean["Year"] = df_clean["OrderDate"].dt.year
df_clean["Quarter"] = df_clean["OrderDate"].dt.quarter
df_clean["Month"] = df_clean["OrderDate"].dt.month
df_clean["MonthName"] = df_clean["OrderDate"].dt.month_name()
df_clean["Day"] = df_clean["OrderDate"].dt.day
df_clean["Week"] = df_clean["OrderDate"].dt.isocalendar().week

df_clean.head()
```

```python
# Mean and median of important metrics

print("Average Revenue:", df_clean["Revenue"].mean())
print("Median Revenue:", df_clean["Revenue"].median())
print("Average Quantity:", df_clean["Quantity"].mean())
print("Median Quantity:", df_clean["Quantity"].median())
```

```python
# IQR outlier detection

Q1 = df_clean["Revenue"].quantile(0.25)
Q3 = df_clean["Revenue"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Identify potential revenue outliers

outliers = df_clean[
    (df_clean["Revenue"] < lower_bound) |
    (df_clean["Revenue"] > upper_bound)
]

outliers
```

```python
# Investigate potential outliers

outliers[
    [
        "OrderID",
        "CustomerID",
        "Product",
        "Quantity",
        "UnitPrice",
        "Revenue"
    ]
]
```

```python
# Calculate revenue by customer

customer_revenue = (
    df_clean
    .groupby("CustomerID")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

customer_revenue
```

```python
# Create customer value segments

customer_segment = pd.qcut(
    customer_revenue.rank(method="first"),
    q=3,
    labels=[
        "Low Value",
        "Medium Value",
        "High Value"
    ]
)

customer_segment.value_counts()
```

```python
# Monthly revenue

monthly_revenue = (
    df_clean
    .groupby("MonthName")["Revenue"]
    .sum()
)

monthly_revenue
```

```python
# Sort by actual date

monthly_revenue = (
    df_clean
    .groupby(
        df_clean["OrderDate"].dt.to_period("M")
    )["Revenue"]
    .sum()
)

monthly_revenue
```

```python
# Monthly revenue trend
monthly_revenue.plot(
    figsize=(12, 5),
    marker="o"
)
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
# Month-over-month growth

mom_growth = (
    monthly_revenue
    .pct_change()
    * 100
)

mom_growth
```

```python
# Month-over-month growth

mom_growth = (
    monthly_revenue
    .pct_change()
    * 100
)

mom_growth
```

```python
# 3-month rolling average

rolling_revenue = (
    monthly_revenue
    .rolling(3)
    .mean()
)

rolling_revenue
```

```python
# Find highest-revenue region

top_region = (
    df_clean
    .groupby("Region")["Revenue"]
    .sum()
    .idxmax()
)

print("Top region:", top_region)
```

```python
# Calculate business KPIs

total_revenue = df_clean["Revenue"].sum()

total_orders = df_clean["OrderID"].nunique()

total_customers = df_clean["CustomerID"].nunique()

total_quantity = df_clean["Quantity"].sum()

average_order_value = (
    total_revenue / total_orders
)

average_price = df_clean["UnitPrice"].mean()

print("Total Revenue:", total_revenue)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Quantity:", total_quantity)
print("Average Order Value:", average_order_value)
print("Average Unit Price:", average_price)
```
