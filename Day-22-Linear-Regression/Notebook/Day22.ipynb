```python
# Linear Regression: Predict Student Marks Based on Study Hours
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# 1. Create the dataset
df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 30, 40, 50, 60]
})
print("Dataset:")
print(df)
```

```python
# 2. Separate features (X) and target (y)
X = df[["Study_Hours"]]
y = df["Marks"]
```

```python
# 3. Create and train the Linear Regression model
model = LinearRegression()
model.fit(X, y)
```

```python

# 4. Display the learned equation
slope = model.coef_[0]
intercept = model.intercept_
print("\nSlope:", slope)
print("Intercept:", intercept
```

```python
# 5. Predict marks for a new student
new_student = pd.DataFrame({
    "Study_Hours": [6]
})
predicted_marks = model.predict(new_student)
print("\nStudy Hours:", new_student["Study_Hours"].iloc[0])
print("Predicted Marks:", predicted_marks[0])
```

```python
# 6. Predict marks for multiple students
new_students = pd.DataFrame({
    "Study_Hours": [6, 7, 8, 9]
})

new_students["Predicted_Marks"] = model.predict(new_students)

print("\nPredictions for New Students:")
print(new_students)
```

```python
# 7. Visualize the data and regression line
plt.scatter(X["Study_Hours"], y, color="blue", label="Actual Marks")
plt.plot(
    X["Study_Hours"],
    model.predict(X),
    color="red",
    label="Regression Line"
)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Linear Regression: Study Hours vs Marks")
plt.legend()
plt.grid(True)
plt.show()
```
