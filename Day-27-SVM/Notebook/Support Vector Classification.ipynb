```python
# SVC: Student Pass/Fail Prediction
# Import required libraries

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
# Create sample student data
data = {
    "study_hours": [1, 2, 2, 3, 4, 5, 6, 7],
    "attendance": [40, 45, 50, 55, 65, 70, 80, 90],
    "result": [
        "Fail", "Fail", "Fail", "Fail",
        "Pass", "Pass", "Pass", "Pass"
    ]
}

# Convert data into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

```python
# Select input features
X = df[["study_hours", "attendance"]]

# Select the output column
y = df["result"]

# Display inputs and target
print("Features:\n", X)
print("Target:\n", y)
```

```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Display dataset sizes
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# Create a pipeline with scaling and SVC
model = make_pipeline(
    StandardScaler(),
    SVC(kernel="linear", C=1.0)
)

# Display model configuration
print(model)
```

```python
# Create a pipeline with scaling and SVC
model = make_pipeline(
    StandardScaler(),
    SVC(kernel="linear", C=1.0)
)

# Display model configuration
print(model)
```

```python
# Train SVC on the training data
model.fit(X_train, y_train)

# Confirm training is complete
print("SVC model trained successfully!")
```

```python
# Predict results for test students
y_pred = model.predict(X_test)

# Display actual and predicted results
print("Actual results:", list(y_test))
print("Predicted results:", list(y_pred))
```

```python
# Calculate test accuracy
accuracy = accuracy_score(y_test, y_pred)

# Display accuracy
print("Test Accuracy:", accuracy)
print("Test Accuracy (%):", round(accuracy * 100, 2), "%")
```

```python
# Create data for a new student
new_student = pd.DataFrame({
    "study_hours": [5],
    "attendance": [75]
})

# Predict the student's result
prediction = model.predict(new_student)

# Display prediction
print("Study Hours:", new_student["study_hours"].iloc[0])
print("Attendance:", new_student["attendance"].iloc[0], "%")
print("Predicted Result:", prediction[0])
```

```python
# Create data for multiple new students
new_students = pd.DataFrame({
    "study_hours": [2, 4, 6],
    "attendance": [45, 60, 85]
})

# Predict all student results
predictions = model.predict(new_students)

# Display inputs and predictions together
new_students["predicted_result"] = predictions
print(new_students)
```
