```python
# task 1 - Create a Pandas Series
# Create a Series containing 5 student marks

marks = pd.Series([85, 90, 78, 92, 88])

# Display the Series
print(marks)
```

```python
# task 2 - Create a DataFrame
# Create a DataFrame containing student names, ages and marks

data = {
    "Name": ["Gaurav", "Rahul", "Amit", "Sneha", "Priya"],
    "Age": [22, 23, 21, 22, 24],
    "Marks": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
# task 3 - Display DataFrame Information
# Display the shape, columns and data types

print("Shape:", df.shape)

print("Columns:")
print(df.columns)

print("Data Types:")
print(df.dtypes)
```

```python

# Select and display the Name and Marks columns

selected_columns = df[["Name", "Marks"]]

# Display selected columns
print(selected_columns)
```

```python
# Display the first and third rows using loc

first_row = df.loc[0]
third_row = df.loc[2]

# Display the rows
print("First Row:")
print(first_row)

print("Third Row:")
print(third_row)
```

```python
# Display the first three rows using iloc

selected_rows = df.iloc[0:3]

# Display the selected rows
print(selected_rows)
```

```python
# Display students who scored more than 85 marks

filtered_data = df[df["Marks"] > 85]

# Display filtered data
print(filtered_data)
```

```python
# Sort students according to marks in descending order

sorted_df = df.sort_values("Marks", ascending=False)

# Display the sorted DataFrame
print(sorted_df)
```

```python
# Create a Result column based on student marks

df["Result"] = df["Marks"] >= 40

# Display the updated DataFrame
print(df)
```

```python
# Calculate the sum, average, maximum and minimum marks

print("Total Marks:", df["Marks"].sum())
print("Average Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())
```

```python
print(df.head())
```

```python
# Display statistical summary

print(df.describe())
```

```python
# Find the student with the highest marks

highest_marks = df["Marks"].max()

print("Highest Marks:", highest_marks)
```

```python
# Find the student with the lowest marks

lowest_marks = df["Marks"].min()

print("Lowest Marks:", lowest_marks)
```

```python
# Find the average age of students

average_age = df["Age"].mean()

print("Average Age:", average_age)
```

```python
# Display unique values from the Result column

print(df["Result"].unique())
```

```python
# Count how many students passed and failed

print(df["Result"].value_counts())
```
