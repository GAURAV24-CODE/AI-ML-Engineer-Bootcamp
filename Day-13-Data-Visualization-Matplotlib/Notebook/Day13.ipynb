```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Topic: First Line Plot
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 40]
plt.plot(x, y)
plt.show()
```

```python
# Topic: Line Plot
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 12000, 15000, 13000, 18000]
plt.plot(months, sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales Data")
plt.show()
```

```python
# Topic: Markers and Line Styles
plt.plot(
    months,
    sales,
    marker="*",
    linestyle="--",
    linewidth=2
)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# Topic: Bar Chart
products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales_units = [500, 800, 300, 650]

plt.bar(products, sales_units)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.show()
```

```python
# Topic: Horizontal Bar Chart

plt.barh(products, sales_units, color='green')

plt.title("Product Sales")
plt.xlabel("Units Sold")
plt.ylabel("Product")

plt.show()
```

```python
# Topic: Histogram

ages = [
    18, 20, 21, 22, 22,
    23, 24, 25, 26, 27,
    28, 30, 31, 32, 35,
    36, 38, 40, 42, 45
]

plt.hist(ages, bins=10, color='blue', edgecolor='black')

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
# Topic: Scatter Plot

hours = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 40, 45, 55, 65, 72, 80]

plt.scatter(hours, marks, color='red', marker='o')

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# Topic: Pie Chart

skills = ["Python", "SQL", "Machine Learning", "Statistics"]
hours = [40, 25, 20, 15]

plt.pie(
    hours,
    labels=skills,
    autopct="%1.1f%%"
)

plt.title("Learning Time Distribution")

plt.show()
```

```python
# Topic: Box Plot

salary = [
    25000,
    28000,
    30000,
    32000,
    35000,
    36000,
    40000,
    150000
]

plt.boxplot(salary, vert=True)

plt.title("Salary Distribution")
plt.ylabel("Salary")

plt.show()
```

```python
# Topic: Titles and Axis Labels

months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [10000, 12000, 15000, 13000, 18000]

plt.plot(months, revenue)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
```

```python
# Topic: Legend

months = ["Jan", "Feb", "Mar", "Apr", "May"]

laptop = [100, 120, 150, 140, 180]
mobile = [150, 160, 170, 190, 210]

plt.plot(months, laptop, label="Laptop")
plt.plot(months, mobile, label="Mobile")

plt.title("Product Sales")
plt.xlabel("Month")
plt.ylabel("Units Sold")

plt.legend()

plt.show()
```

```python
# Topic: Grid

plt.plot(
    months,
    laptop,
    marker="o"
)

plt.title("Laptop Sales")
plt.xlabel("Month")
plt.ylabel("Units Sold")

plt.grid()

plt.show()
```

```python
# Topic: Multiple Data Series

months = ["Jan", "Feb", "Mar", "Apr", "May"]

laptop = [100, 120, 150, 140, 180]
mobile = [150, 160, 170, 190, 210]
tablet = [80, 90, 100, 95, 120]

plt.plot(months, laptop, label="Laptop")
plt.plot(months, mobile, label="Mobile")
plt.plot(months, tablet, label="Tablet")

plt.title("Monthly Product Sales")
plt.xlabel("Month")
plt.ylabel("Units Sold")

plt.legend()
plt.grid()

plt.show()
```

```python
# Topic: Subplots

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(months, laptop)
axes[0, 0].set_title("Line Plot")

axes[0, 1].bar(products, sales_units)
axes[0, 1].set_title("Bar Chart")

axes[1, 0].hist(ages)
axes[1, 0].set_title("Histogram")

axes[1, 1].scatter(hours, marks)
axes[1, 1].set_title("Scatter Plot")

plt.tight_layout()

plt.show()
```

```python
# Topic: Figure Size

plt.figure(figsize=(10, 5))

plt.plot(months, revenue)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
```

```python
# Topic: Save Figure

plt.plot(months, revenue)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.savefig("monthly_revenue.png", dpi=300)

plt.show()
```

```python
# Topic: Import Pandas and Matplotlib

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Topic: Create DataFrame

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [10000, 12000, 15000, 13000, 18000, 21000]
}

df = pd.DataFrame(data)

df
```

```python
# Topic: Plot Pandas DataFrame

plt.plot(
    df["Month"],
    df["Sales"],
    marker="o"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()

plt.show()
```

```python
# Topic: E-Commerce Dataset

data = {
    "Order_ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Product": [
        "Laptop",
        "Mobile",
        "Laptop",
        "Tablet",
        "Mobile",
        "Headphones",
        "Tablet",
        "Laptop"
    ],
    "Category": [
        "Electronics",
        "Electronics",
        "Electronics",
        "Electronics",
        "Electronics",
        "Accessories",
        "Electronics",
        "Electronics"
    ],
    "Sales": [
        70000,
        30000,
        65000,
        25000,
        35000,
        8000,
        28000,
        72000
    ],
    "Quantity": [2, 5, 1, 4, 6, 10, 3, 2],
    "Profit": [10000, 5000, 8000, 4000, 6000, 2000, 4500, 12000]
}

df = pd.DataFrame(data)

df
```

```python
# Topic: Dataset Information

df.info()
```

```python
# Topic: Descriptive Statistics

df.describe()
```

```python
# Topic: Sales by Product

product_sales = df.groupby("Product")["Sales"].sum()

product_sales
```

```python
# Topic: Product Sales Bar Chart

plt.figure(figsize=(8, 5))

plt.bar(
    product_sales.index,
    product_sales.values,
)

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.show()
```

```python
# Topic: Sales Distribution

plt.hist(
    df["Sales"],
    bins=5
)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
# Topic: Quantity vs Sales

plt.scatter(
    df["Quantity"],
    df["Sales"]
)

plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")

plt.show()
```

```python
# Topic: Profit Distribution

plt.boxplot(df["Profit"])

plt.title("Profit Distribution")
plt.ylabel("Profit")

plt.show()
```
