```python
# Import Pandas
import pandas as pd

# Create a sample dataset
df = pd.DataFrame({
    "Age": [22, 25, 30, 35],
    "Income": [30000, 40000, 55000, 70000],
    "Bedrooms": [1, 2, 3, 4],
    "Price": [30, 45, 65, 90]
})

# Select only the input features
X = df[["Age", "Income", "Bedrooms"]]

# Display the features
X
```

```python
# Select numerical features
numerical_features = df[["Age", "Income", "Bedrooms"]]

# Display numerical features
print(numerical_features)
```

```python
# Add a categorical feature
df["City"] = ["Nashik", "Pune", "Mumbai", "Nashik"]

# Display the categorical feature
print(df["City"])
```

```python
# Create a new feature from existing features
df["Income_per_Age"] = df["Income"] / df["Age"]

# Display the newly created feature
print(df[["Income", "Age", "Income_per_Age"]])
```

```python
# Select only the features we want to use
selected_features = df[["Age", "Income", "Bedrooms"]]

# Display selected features
selected_features
```
