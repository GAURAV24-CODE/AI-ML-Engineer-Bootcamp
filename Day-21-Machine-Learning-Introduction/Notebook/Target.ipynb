```python
# Import Pandas
import pandas as pd

# Create a sample regression dataset
df = pd.DataFrame({
    "Area": [1000, 1500, 2000, 2500],
    "Bedrooms": [2, 3, 4, 5],
    "Price": [50, 75, 100, 125]
})

# Select input features
X = df[["Area", "Bedrooms"]]

# Select the target variable
y = df["Price"]

# Display X and y
print("Features (X):")
print(X)

print("\nTarget (y):")
print(y)
```

```python
# Create a simple classification dataset
df = pd.DataFrame({
    "Age": [20, 25, 30, 35],
    "Income": [25000, 35000, 50000, 70000],
    "Purchased": ["No", "No", "Yes", "Yes"]
})

# Select features
X = df[["Age", "Income"]]

# Select classification target
y = df["Purchased"]

# Display the target
print("Classification Target:")
print(y)
```

```python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression

# Create the classification model
model = LogisticRegression()

# Train using features and target
model.fit(X, y)

# Predict a new customer
prediction = model.predict([[28, 45000]])

# Display prediction
print("Prediction:", prediction[0])
```
