# 🌳 Decision Tree — Tuning Cheat Sheet

### 1. Important Variables

| Variable | Purpose |
|---|---|
| `X` | Input features |
| `y` | Target/output labels |
| `test_size=0.25` | Uses 25% data for testing |
| `random_state=42` | Ensures reproducible results |
| `stratify=y` | Maintains class proportions |
| `y_pred` | Stores predicted labels |

### 2. Important Hyperparameters

| Hyperparameter | Why Use It? |
|---|---|
| `max_depth` | Limits tree depth to help prevent overfitting |
| `min_samples_split` | Minimum samples required to split a node |
| `min_samples_leaf` | Minimum samples required in a leaf node |
| `criterion` | Measures split quality (`gini`, `entropy`, `log_loss`) |
| `max_features` | Controls features considered at each split |
| `class_weight` | Helps handle imbalanced classes |

### 3. Model Training & Tuning

| Variable / Word | Purpose |
|---|---|
| `DecisionTreeClassifier()` | Creates a classification tree |
| `param_grid` | Stores hyperparameter values to test |
| `GridSearchCV()` | Searches parameter combinations using cross-validation |
| `cv=4` | Uses 4 cross-validation folds |
| `scoring="accuracy"` | Uses accuracy to compare models |
| `n_jobs=-1` | Runs the search in parallel using available CPU cores |
| `best_params_` | Best hyperparameter combination |
| `best_score_` | Best mean cross-validation score |
| `best_estimator_` | Best fitted model |

### 4. Evaluation & Key Terms

| Term | Meaning |
|---|---|
| Accuracy | Proportion of correct predictions |
| Precision | Correct positive predictions among predicted positives |
| Recall | Correct positive predictions among actual positives |
| F1-score | Balance between precision and recall |
| Gini impurity | Measures how mixed the classes are in a node |
| Entropy | Measures class disorder in a node |
| Overfitting | Tree learns training data too closely |
| Pruning | Reduces tree complexity |
| Feature importance | Estimates each feature's contribution to tree decisions |
| Cross-validation | Evaluates performance across multiple data folds |

### 5. Essential Workflow

**Prepare `X` and `y` → Split Data → Train Decision Tree → Tune with `GridSearchCV` → Evaluate on Test Data**

**Remember:** Limit tree depth and tune minimum sample sizes to help reduce overfitting. Fit and tune using training data only; reserve test data for final evaluation.

```python
# Cell 1: Import required libraries

import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Cell 2: Create a simple fruit dataset

data = {
    "Weight": [
        150, 160, 170, 140, 155, 165, 175, 145,
        130, 135, 125, 138, 145, 128, 132, 140,
        180, 190, 185, 195, 120, 118, 122, 115
    ],
    "Sweetness": [
        7, 8, 7, 6, 8, 7, 9, 6,
        4, 5, 3, 4, 5, 3, 4, 5,
        8, 9, 8, 9, 3, 2, 3, 2
    ],
    "Fruit": [
        "Apple", "Apple", "Apple", "Apple",
        "Apple", "Apple", "Apple", "Apple",
        "Orange", "Orange", "Orange", "Orange",
        "Orange", "Orange", "Orange", "Orange",
        "Apple", "Apple", "Apple", "Apple",
        "Orange", "Orange", "Orange", "Orange"
    ]
}

df = pd.DataFrame(data)

print(df.head())
print("\nDataset shape:", df.shape)
print("\nClass counts:\n", df["Fruit"].value_counts())
```

```python
# Cell 3: Separate features and targeeeeeeeeeeet

X = df[["Weight", "Sweetness"]]
y = df["Fruit"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python

# Cell 4: Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# Cell 5: Train the baseline Decision Tree

baseline_model = DecisionTreeClassifier(random_state=42)

baseline_model.fit(X_train, y_train)

baseline_predictions = baseline_model.predict(X_test)

baseline_accuracy = accuracy_score(
    y_test,
    baseline_predictions
)

print("Baseline Accuracy:", baseline_accuracy)
```

```python
# Cell 6: Tune max_depth only

for depth in [1, 2, 3, 5, None]:
    model = DecisionTreeClassifier(
        max_depth=depth,
        random_state=42
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    score = accuracy_score(y_test, predictions)

    print(f"max_depth={depth}: Accuracy={score:.2f}")
```

```python
# Cell 7: Tune min_samples_split only

for split_value in [2, 4, 6, 8, 10]:
    model = DecisionTreeClassifier(
        min_samples_split=split_value,
        random_state=42
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    score = accuracy_score(y_test, predictions)

    print(
        f"min_samples_split={split_value}: "
        f"Accuracy={score:.2f}"
    )
```

```python
# Cell 8: Tune min_samples_leaf only

for leaf_value in [1, 2, 3, 4, 5]:
    model = DecisionTreeClassifier(
        min_samples_leaf=leaf_value,
        random_state=42
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    score = accuracy_score(y_test, predictions)

    print(
        f"min_samples_leaf={leaf_value}: "
        f"Accuracy={score:.2f}"
    )
```

```python
# Cell 10: Compare max_features

for feature_value in [None, "sqrt", "log2"]:
    model = DecisionTreeClassifier(
        max_features=feature_value,
        random_state=42
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    score = accuracy_score(y_test, predictions)

    print(
        f"max_features={feature_value}: "
        f"Accuracy={score:.2f}"
    )
```

```python
# Cell 9: Compare splitting criteria

for criterion_value in ["gini", "entropy", "log_loss"]:
    model = DecisionTreeClassifier(
        criterion=criterion_value,
        random_state=42
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    score = accuracy_score(y_test, predictions)

    print(
        f"criterion={criterion_value}: "
        f"Accuracy={score:.2f}"
    )
```

```python
# Cell 11: Define the Decision Tree parameter grid

param_grid = {
    "max_depth": [2, 3, 5, None],
    "min_samples_split": [2, 4, 6],
    "min_samples_leaf": [1, 2, 3],
    "criterion": ["gini", "entropy"],
    "max_features": [None, "sqrt"]
}
```

```python
# Cell 12: Find the best hyperparameters

grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    cv=4,
    scoring="accuracy",
    n_jobs=-1,
    refit=True
)

grid_search.fit(X_train, y_train)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest CV Accuracy:")
print(grid_search.best_score_)
```

```python
# Cell 13: Evaluate the best model on test data

best_model = grid_search.best_estimator_

tuned_predictions = best_model.predict(X_test)

tuned_accuracy = accuracy_score(
    y_test,
    tuned_predictions
)

print("Tuned Test Accuracy:", tuned_accuracy)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        tuned_predictions,
        zero_division=0
    )
)
```

```python
# Cell 14: Compare baseline and tuned accuracy

print(f"Baseline Accuracy: {baseline_accuracy:.2%}")
print(f"Tuned Accuracy:    {tuned_accuracy:.2%}")

difference = tuned_accuracy - baseline_accuracy

print(
    f"Accuracy Difference: "
    f"{difference * 100:+.2f} percentage points"
)
```

```python
# Cell 15: Predict a new fruit

new_fruit = pd.DataFrame({
    "Weight": [155],
    "Sweetness": [7]
})

prediction = best_model.predict(new_fruit)

print("Predicted fruit:", prediction[0])
```
