# 🛒 Real-World Application — E-Commerce Analysis

In real Data Science projects, Seaborn can be used to analyze E-Commerce data.

### Business Questions

- Which category has the highest sales?
- Which region performs best?
- What is the sales distribution?
- Are sales related to profit?
- Are there unusual values?
- Which category needs improvement?

### Visualization Strategy

Category → Bar Plot

Sales Distribution → Histogram

Sales vs Profit → Scatter Plot

Outliers → Box Plot

Correlation → Heatmap

### Goal

Convert visualization into actionable business insights.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Topic: Create E-Commerce Dataset
data = {
    "Category": ["Electronics", "Clothing", "Electronics", "Furniture", "Clothing", "Furniture"],
    "Sales": [1200, 800, 1500, 900, 700, 1100],
    "Profit": [200, 150, 300, 100, 120, 180],
    "Region": ["North", "South", "East", "West", "North", "East"]
}

ecommerce = pd.DataFrame(data)
ecommerce
```

```python
# Topic: E-Commerce Category Sales
sns.barplot(data=ecommerce, x="Category", y="Sales", errorbar=None)
plt.title("Average Sales by Category")
plt.show()
```

```python
# Topic: E-Commerce Sales Distribution
sns.histplot(data=ecommerce, x="Sales", bins=5, kde=True)
plt.title("Sales Distribution")
plt.show()
```

```python
# Topic: E-Commerce Sales Distribution
sns.histplot(data=ecommerce, x="Sales", bins=5, kde=True)
plt.title("Sales Distribution")
plt.show()
```

```python
# Topic: Sales and Profit Relationship
sns.scatterplot(data=ecommerce, x="Sales", y="Profit", hue="Category")
plt.title("Sales vs Profit")
plt.show()
```
