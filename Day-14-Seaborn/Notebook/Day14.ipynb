```python
# Topic: Load Seaborn Dataset
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
df = sns.load_dataset("tips")
df.head()
```

```python
#  Dataset Shape
df.shape
```

```python
#  Dataset Shape
df.shape
```

```python
#Statistical Summary
df.describe()
```

```python
# Topic: Seaborn Line Plot
plt.figure(figsize=(13, 5))
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 135, 180, 210]
sns.lineplot(x=months, y=sales)
plt.show()
```

```python
# Topic: Seaborn Bar Plot
plt.figure(figsize=(13, 4))
sns.barplot(data=df, x="day", y="total_bill")
plt.show()
```

```python
# Topic: Seaborn Count Plot
plt.figure(figsize=(13, 4))
sns.countplot(data=df, x="day")
plt.show()
```

```python
# Topic: Seaborn Histogram
plt.figure(figsize=(13, 6))
sns.histplot(data=df, x="total_bill", bins=10)
plt.show()
```

```python
# Topic: KDE Plot
plt.figure(figsize=(13, 4))
sns.kdeplot(data=df, x="total_bill")
plt.show()
```

```python
# Topic: Histogram with KDE
plt.figure(figsize=(13, 4))
sns.histplot(data=df, x="total_bill", kde=True)
plt.show()
```

```python
# Topic: Seaborn Box Plot
plt.figure(figsize=(13, 4))
sns.boxplot(data=df, x="day", y="total_bill")
plt.show()
```

```python
# Topic: Seaborn Violin Plot
plt.figure(figsize=(13, 4))
sns.violinplot(data=df, x="day", y="total_bill")
plt.show()
```

```python
# Topic: Seaborn Strip Plot
plt.figure(figsize=(13, 4))
sns.stripplot(data=df, x="day", y="total_bill", jitter=True)
plt.show()
```

```python
# Topic: Seaborn Swarm Plot
plt.figure(figsize=(13, 4))
sns.swarmplot(data=df, x="day", y="total_bill", hue="sex")
plt.show()
```

```python
# Topic: Seaborn Scatter Plot
sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex")
plt.show()
```

```python
# Topic: Scatter Plot with Style
sns.scatterplot(data=df, x="total_bill", y="tip", style="time", hue="sex")
plt.show()
```

```python
# Topic: Regression Plot


sns.regplot(data=df, x="total_bill", y="tip", ci=None)
plt.show()
```

```python
# Topic: Correlation Heatmap
plt.figure(figsize=(13, 4))
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()
sns.heatmap(correlation, annot=True)
plt.show()
```

```python
# Topic: Pair Plot
plt.figure(figsize=(13, 4))
sns.pairplot(df)
plt.show()
```

```python
# Topic: Seaborn Color Palette
plt.figure(figsize=(13, 4))
sns.countplot(data=df, x="day", hue="day", palette="Set2", legend=False)
plt.show()
```

```python
# Topic: Plot Labels
plt.figure(figsize=(13, 4))
sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex")
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# Topic: Seaborn with Matplotlib
plt.figure(figsize=(13, 5))
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="day", y="total_bill")
plt.title("Total Bill Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```
