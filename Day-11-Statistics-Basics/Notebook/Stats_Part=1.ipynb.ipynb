```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Create a sample student performance dataset

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Study_Hours": [2, 3, 4, 5, 5, 6, 7, 8, 9, 10],
    "Attendance": [65, 70, 75, 80, 82, 85, 88, 90, 92, 95],
    "Marks": [45, 50, 55, 60, 65, 68, 72, 78, 85, 90]
}

df = pd.DataFrame(data)

df
```

```python
# Check basic dataset information
print("--------------------------------")
print("Shape:", df.shape)
print("--------------------------------")
print("\nColumns:")
print(df.columns.tolist())
print("--------------------------------")

print("\nData Types:")
print(df.dtypes)
```

```python
print("\nDataset Information:")
print("--------------------------------")
df.info()
```

```python
# Generate descriptive statistics
display(df.describe())
```

```python
# Calculate central tendency

print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Mode Marks:", df["Marks"].mode().tolist())
```

```python
# Calculate range
marks_range = df["Marks"].max() - df["Marks"].min()
print("Marks Range:", marks_range)
```

```python
# Calculate variance
variance = df["Marks"].var()
print("Variance:", variance)
```

```python
# Calculate standard deviation
std = df["Marks"].std()
print("Standard Deviation:", std)
```

```python
# Calculate quartiles

Q1 = df["Marks"].quantile(0.25)
Q2 = df["Marks"].quantile(0.50)
Q3 = df["Marks"].quantile(0.75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

```python
# Calculate selected percentiles
print("10th Percentile:", df["Marks"].quantile(0.10))
print("25th Percentile:", df["Marks"].quantile(0.25))
print("50th Percentile:", df["Marks"].quantile(0.50))
print("75th Percentile:", df["Marks"].quantile(0.75))
print("90th Percentile:", df["Marks"].quantile(0.90))
```

```python
# Calculate Interquartile Range
IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
# Detect potential outliers using IQR
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
outliers = df[
    (df["Marks"] < lower_bound) |
    (df["Marks"] > upper_bound)
]
print("\nPotential Outliers:")
display(outliers)
```

```python
# Visualize marks distribution
sns.histplot(df["Marks"],color="green", kde=True)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# Visualize spread and outliers
sns.boxplot(x=df["Marks"])

plt.title("Marks Boxplot")
plt.xlabel("Marks")
plt.show()
```

```python
# Calculate skewness
skewness = df["Marks"].skew()
print("Skewness:", skewness)
```

```python
# Calculate kurtosis

kurtosis = df["Marks"].kurt()

print("Kurtosis:", kurtosis)
```

```python
# Calculate Z-score manually

df["Z_Score"] = (
    (df["Marks"] - df["Marks"].mean())
    / df["Marks"].std()
)

df[["Student", "Marks", "Z_Score"]]
```

```python
# Calculate covariance between variables

covariance = df[
    ["Study_Hours", "Attendance", "Marks"]
].cov()

covariance
```

```python
# Calculate correlation

correlation = df[
    ["Study_Hours", "Attendance", "Marks"]
].corr()

correlation
```

```python
# Visualize correlations
plt.figure(figsize=(12, 5))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Study Hours vs Marks

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.title("Study Hours vs Marks")
plt.show()
```

```python
# Compare statistics of important numerical columns

statistics = df[
    ["Study_Hours", "Attendance", "Marks"]
].agg([
    "mean",
    "median",
    "std",
    "min",
    "max"
])

statistics
```

```python
# Final summary

print("===== STATISTICAL SUMMARY =====")

print("Mean Marks:", round(df["Marks"].mean(), 2))
print("Median Marks:", round(df["Marks"].median(), 2))
print("Standard Deviation:", round(df["Marks"].std(), 2))
print("Variance:", round(df["Marks"].var(), 2))
print("IQR:", round(IQR, 2))
print("Skewness:", round(skewness, 2))
print("Kurtosis:", round(kurtosis, 2))

print("\nCorrelation:")
print(correlation)
```
