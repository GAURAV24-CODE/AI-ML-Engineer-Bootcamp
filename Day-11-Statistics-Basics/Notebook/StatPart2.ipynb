```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

print("Libraries imported!")
```

```python
# Create population
population = np.arange(1, 1001)

# Select 50 people as sample
sample = np.random.choice(population, 50, replace=False)

print("Population:", len(population))
print("Sample:", len(sample))
```

```python
# Create population and sample
population = np.random.normal(50, 10, 10000)
sample = np.random.choice(population, 100)

# Calculate means
print("Population Mean:", population.mean())
print("Sample Mean:", sample.mean())
```

```python
## 3. Simple Random Sampling
# # Population of 100 students
students = np.arange(1, 101)

# Randomly select 10 students
sample = np.random.choice(students, 10, replace=False)

print("Sample:", sample)
```

```python
# Create population
population = np.random.normal(50, 10, 10000)

# Calculate 1,000 sample means
means = [np.mean(np.random.choice(population, 30))
         for _ in range(1000)]

print("Mean of sample means:", np.mean(means))
```

```python
# Sample data
data = [45, 50, 52, 48, 55, 60, 51, 49, 53, 47]

# Calculate standard error
se = stats.sem(data)

print("Standard Error:", se)
```

```python
## 6. Central Limit Theorem (CLT)
# Create a skewed population
population = np.random.exponential(10, 10000)

# Generate sample means
means = [np.mean(np.random.choice(population, 50))
         for _ in range(1000)]

# Plot
plt.hist(means, bins=30)
plt.title("Central Limit Theorem")
plt.show()
```

```python
## 7. Point Estimation
# Customer spending sample
spending = [800, 900, 750, 1000, 850, 700, 950]

# Point estimate
estimate = np.mean(spending)

print("Estimated Average:", estimate)
```

```python
## 8. Confidence Interval
# Sample data
data = [55, 60, 62, 58, 65, 70, 68, 72, 64, 61]

# Calculate 95% confidence interval
ci = stats.t.interval(
    0.95,
    len(data) - 1,
    loc=np.mean(data),
    scale=stats.sem(data)
)

print("Mean:", np.mean(data))
print("95% CI:", ci)
```

```python
# Calculate margin of error
mean = np.mean(data)
margin = ci[1] - mean

print("Margin of Error:", margin)
```

```python
## 10. Hypothesis Testing
# Example p-value
p = 0.03
alpha = 0.05

# Make decision
print("Reject H₀" if p <= alpha else "Fail to Reject H₀")
```

```python
## 11. Z-Test
# Given values
sample_mean = 503
population_mean = 500
population_sd = 10
n = 100

# Calculate Z and p-value
z = (sample_mean - population_mean) / (population_sd / np.sqrt(n))
p = 2 * (1 - stats.norm.cdf(abs(z)))

print("Z:", z)
print("P-value:", p)
print("Reject H₀" if p <= 0.05 else "Fail to Reject H₀")
```

```python
## 12. One-Sample T-Test# Student scores
scores = [55, 62, 61, 58, 65, 63, 59, 64, 60, 66]

# One-sample T-test
t, p = stats.ttest_1samp(scores, 60)

print("T:", t)
print("P-value:", p)
print("Reject H₀" if p <= 0.05 else "Fail to Reject H₀")
```

```python
## 13. Independent T-Test
# Sales of two independent teams
A = [70, 72, 68, 75, 71]
B = [78, 80, 76, 82, 79]

# Independent T-test
t, p = stats.ttest_ind(A, B)

print("T:", t)
print("P-value:", p)
print("Reject H₀" if p <= 0.05 else "Fail to Reject H₀")
```

```python
## 14. Paired T-Test
# Same students before and after training
before = [60, 62, 58, 65, 61]
after = [68, 70, 67, 72, 69]

# Paired T-test
t, p = stats.ttest_rel(before, after)

print("T:", t)
print("P-value:", p)
print("Reject H₀" if p <= 0.05 else "Fail to Reject H₀")
```

```python
## 15. One-Way ANOVA
# Three groups
A = [50, 52, 48, 51, 49]
B = [55, 57, 54, 56, 58]
C = [70, 72, 68, 71, 73]

# ANOVA test
f, p = stats.f_oneway(A, B, C)

print("F:", f)
print("P-value:", p)
print("Reject H₀" if p <= 0.05 else "Fail to Reject H₀")
```

```python
## 16. Chi-Square Goodness-of-Fit

### Definition Checks whether observed frequencies match expected frequencies.
# Observed and expected frequencies
observed = [30, 25, 20, 25]
expected = [25, 25, 25, 25]

# Chi-square test
chi, p = stats.chisquare(observed, f_exp=expected)

print("Chi-square:", chi)
print("P-value:", p)
print("Reject H₀" if p <= 0.05 else "Fail to Reject H₀")
```

```python
## 17. Chi-Square Test of Independence

### Definition Checks whether two categorical variables are related.
# Contingency table
table = [
    [40, 20, 10],
    [25, 35, 15],
    [15, 20, 30]
]

# Chi-square independence test
chi, p, dof, expected = stats.chi2_contingency(table)

print("Chi-square:", chi)
print("P-value:", p)
print("Reject H₀" if p <= 0.05 else "Fail to Reject H₀")
```

```python
## 19. Normality Check

### Definition Checks whether data approximately follows a normal distribution.
# Sample data
data = [12, 15, 14, 16, 13, 15, 17, 14, 16, 15]

# Shapiro-Wilk test
stat, p = stats.shapiro(data)

print("P-value:", p)
print("Normal" if p > 0.05 else "Not Normal")
```

```python
## 26. A/B Testing

### Definition A/B testing compares two versions to determine which performs better
from statsmodels.stats.proportion import proportions_ztest

# [conversions, visitors]
A = [100, 1000]
B = [130, 1000]

# Compare conversion rates
z, p = proportions_ztest(
    [A[0], B[0]],
    [A[1], B[1]]
)

print("A Rate:", A[0] / A[1])
print("B Rate:", B[0] / B[1])
print("P-value:", p)

print("Significant Difference"
      if p <= 0.05
      else "No Significant Difference")
```
