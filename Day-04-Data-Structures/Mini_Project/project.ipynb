```python
# ============================================================
# 🎓 Student Record Management System
# Day 4 Mini Project - AI/ML Engineer Bootcamp
#
# Concepts Used:
# ✅ Lists
# ✅ Dictionaries
# ✅ Functions
# ✅ Loops
# ✅ Conditional Statements
# Features:
# 1. Add Student
# 2. Display Students
# 3. Search Student
# 4. Update Student Marks
# 5. Delete Student
# 6. Exit
# ============================================================

# List to store student records
students = []

# ============================================================
# Function to Add Student
# ============================================================

def add_student():
    """Add a new student to the record."""

    # Take input from user
    name = input("Enter Student Name: ")
    marks = int(input("Enter Student Marks: "))
    course = input("Enter Course: ")

    # Create dictionary
    student = {
        "Name": name,
        "Marks": marks,
        "Course": course
    }

    # Add dictionary to list
    students.append(student)

    print("\n✅ Student Added Successfully!")


# ============================================================
# Function to Display Students
# ============================================================

def display_students():
    """Display all student records."""

    if len(students) == 0:
        print("\n❌ No Student Records Found.")
        return

    print("\n========== STUDENT RECORDS ==========")

    for index, student in enumerate(students, start=1):
        print(f"\nStudent {index}")
        print(f"Name   : {student['Name']}")
        print(f"Marks  : {student['Marks']}")
        print(f"Course : {student['Course']}")


# ============================================================
# Function to Search Student
# ============================================================

def search_student():
    """Search student using name."""

    search_name = input("Enter Student Name: ")

    for student in students:

        if student["Name"].lower() == search_name.lower():

            print("\n✅ Student Found")

            print("----------------------------")
            print("Name   :", student["Name"])
            print("Marks  :", student["Marks"])
            print("Course :", student["Course"])

            return

    print("\n❌ Student Not Found.")


# ============================================================
# Function to Update Marks
# ============================================================

def update_marks():
    """Update marks of an existing student."""

    search_name = input("Enter Student Name: ")

    for student in students:

        if student["Name"].lower() == search_name.lower():

            new_marks = int(input("Enter New Marks: "))

            student["Marks"] = new_marks

            print("\n✅ Marks Updated Successfully!")

            return

    print("\n❌ Student Not Found.")


# ============================================================
# Function to Delete Student
# ============================================================

def delete_student():
    """Delete a student record."""

    search_name = input("Enter Student Name: ")

    for student in students:

        if student["Name"].lower() == search_name.lower():

            students.remove(student)

            print("\n✅ Student Deleted Successfully!")

            return

    print("\n❌ Student Not Found.")


# ============================================================
# Main Menu
# ============================================================

while True:

    print("\n===================================")
    print("🎓 STUDENT RECORD MANAGEMENT SYSTEM")
    print("===================================")

    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Update Marks")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("\nEnter Your Choice (1-6): ")

    # Option 1
    if choice == "1":
        add_student()

    # Option 2
    elif choice == "2":
        display_students()

    # Option 3
    elif choice == "3":
        search_student()

    # Option 4
    elif choice == "4":
        update_marks()

    # Option 5
    elif choice == "5":
        delete_student()

    # Option 6
    elif choice == "6":
        print("\n🎉 Thank You for Using Student Record Management System!")
        print("Happy Coding! 🚀")
        break

    # Invalid Choice
    else:
        print("\n❌ Invalid Choice! Please enter a number between 1 and 6.")
```
