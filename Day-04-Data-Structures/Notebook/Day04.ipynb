# 📌 What is a Data Structure?

A Data Structure is a way to organize and store data efficiently.

Python provides four built-in data structures:

1. List
2. Tuple
3. Set
4. Dictionary

```python
# Accessing list elements using indexing

# First element
print("First Fruit :", fruits[0])

# Second element
print("Second Fruit :", fruits[1])

# Last element using negative indexing
print("Last Fruit :", fruits[-1])
```

# 📌 Adding Elements to a List

Python provides several methods to add elements.

- append() → Adds an element at the end.
- insert() → Adds an element at a specific position.

```python
# Add a new fruit at the end

fruits.append("Orange")

# Insert a fruit at index 1

fruits.insert(1, "Grapes")

# Display the updated list

print(fruits)
```

# 📌 Removing Elements from a List

Common methods:

- remove()
- pop()
- clear()

```python
# Remove Banana from the list

fruits.remove("Banana")

# Display updated list

print(fruits)
```

# 📌 Sorting Lists

sort() sorts the list in ascending order.

```python
# Create a list of numbers

numbers = [5, 2, 8, 1, 9]

# Sort the list

numbers.sort()

# Print sorted list

print(numbers)
```

```python
# Sort in descending order

numbers.sort(reverse=True)

print(numbers)
```

# 📌 Tuples

Tuple is an ordered and immutable collection.

Immutable means the values cannot be changed after creation.

```python
# Create a tuple

marks = (78, 85, 91, 67)

# Print tuple

print(marks)

# Print data type

print(type(marks))
```

```python
# Access tuple elements

print("First Marks :", marks[0])

print("Last Marks :", marks[-1])
```

# 📌 Sets

A Set stores only unique values.

Duplicate values are automatically removed.

```python
# Create a set

colors = {"Red", "Blue", "Green", "Red"}

# Print the set

print(colors)
```

```python
# Create two sets

A = {1, 2, 3}

B = {3, 4, 5}

# Union

print("Union :", A.union(B))

# Intersection

print("Intersection :", A.intersection(B))

# Difference

print("Difference :", A.difference(B))
```

# 📌 Dictionaries

Dictionary stores data as Key : Value pairs.

```python
# Create a dictionary

student = {
    "Name": "Gaurav",
    "Age": 22,
    "Course": "MCA"
}

# Print dictionary

print(student)
```

```python
# Access values using keys

print(student["Name"])

# Add a new key-value pair

student["City"] = "Pune"

# Print updated dictionary

print(student)
```

```python
# Print all keys

print(student.keys())

# Print all values

print(student.values())

# Print all key-value pairs

print(student.items())
```

# 📌 Nested Data Structures

We can combine multiple data structures.

Example:
A list of dictionaries.

```python
# List containing dictionaries

students = [
    {
        "Name": "Rahul",
        "Marks": 89
    },
    {
        "Name": "Amit",
        "Marks": 91
    }
]

# Print all students

print(students)
```
