```python
# ==========================================
# Task 1: Create Your First Class
# ==========================================

# Create an empty class
class Student:
    pass

# Create an object of Student class
student = Student()

# Print the object
print(student)
```

```python
# ==========================================
# Task 2: Class Attributes
# ==========================================

# Create Student class
class Student:

    # Class attributes
    name = "Gaurav"
    age = 22
    course = "AI/ML"

# Create object
student = Student()

# Print attributes
print("Name :", student.name)
print("Age :", student.age)
print("Course :", student.course)
```

```python
# ==========================================
# Task 3: Constructor
# ==========================================

class Student:

    # Constructor
    def __init__(self, name, age):

        # Store values
        self.name = name
        self.age = age

# Create object
student = Student("Gaurav", 22)

# Print values
print(student.name)
print(student.age)
```

```python
# ==========================================
# Task 4: Methods
# ==========================================

class Student:

    def __init__(self, name):
        self.name = name

    # Create method
    def display(self):
        print("Student Name:", self.name)

student = Student("Gaurav")

# Call method
student.display()
```

```python
# ==========================================
# Task 5: Instance Variables
# ==========================================

class Laptop:

    def __init__(self, brand, price):

        # Instance variables
        self.brand = brand
        self.price = price

# Create objects
lap1 = Laptop("Dell", 60000)
lap2 = Laptop("HP", 55000)

# Print details
print(lap1.brand, lap1.price)
print(lap2.brand, lap2.price)
```

```python
# ==========================================
# Task 6: Class Variables
# ==========================================

class Student:

    # Class variable
    college = "SSBT"

    def __init__(self, name):
        self.name = name

s1 = Student("Gaurav")
s2 = Student("Rahul")

# Print class variable
print(s1.college)
print(s2.college)
```

```python
# ==========================================
# Task 7: Inheritance
# ==========================================

# Parent class
class Animal:

    def sound(self):
        print("Animal makes sound")

# Child class
class Dog(Animal):

    def bark(self):
        print("Dog Barks")

dog = Dog()

# Call parent method
dog.sound()

# Call child method
dog.bark()
```

```python
# ==========================================
# Task 8: Polymorphism
# ==========================================

class Dog:

    def sound(self):
        print("Dog Barks")

class Cat:

    def sound(self):
        print("Cat Meows")

# Create list
animals = [Dog(), Cat()]

# Loop through objects
for animal in animals:
    animal.sound()
```

```python
# ==========================================
# Task 9: Encapsulation
# ==========================================

class Bank:

    def __init__(self):

        # Private variable
        self.__balance = 5000

    def show_balance(self):
        print("Balance:", self.__balance)

bank = Bank()

bank.show_balance()
```

```python
# ==========================================
# Task 10: Abstraction
# ==========================================

from abc import ABC, abstractmethod

# Abstract class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

# Child class
class Square(Shape):

    def area(self):
        print("Area of Square")

square = Square()

square.area()
```
