```python
# Probability of getting Head when tossing a fair coin

favorable = 1
total = 2

probability = favorable / total

print("Probability:", probability)
```

```python
import random

# Experiment: Toss a coin
outcome = random.choice(["Head", "Tail"])

print("Outcome:", outcome)
```

```python
# Sample space of a dice roll

sample_space = {1, 2, 3, 4, 5, 6}

print("Sample Space:", sample_space)
print("Total Outcomes:", len(sample_space))
```

```python
# Calculate probability of rolling an even number

even = {2, 4, 6}
total = {1, 2, 3, 4, 5, 6}
p_even = len(even) / len(total)
print("P(Even):", p_even)
```

```python
# Basic Probability Rules
# Addition Rule
# Addition probability
p_a = 0.3
p_b = 0.4
p_ab = 0.1

p_a_or_b = p_a + p_b - p_ab
print("P(A or B):", p_a_or_b)
```

```python
# Multiplication Rule  For independent events:
# P(A and B) = P(A) × P(B)
P_A = 0.5
P_B = 0.5

P_A_and_B = P_A * P_B
print("P(A and B):", P_A_and_B)
```

```python
# Complement probability

p_rain = 0.30

p_no_rain = 1 - p_rain

print("P(No Rain):", p_no_rain)
```

```python
# Conditional probability
### Formul P(A|B) = P(A and B) / P(B)
p_buy = 0.4
p_visit = 0.8
p_buy_and_visit = 0.3

p_buy_given_visit = p_buy_and_visit / p_visit

print("P(Buy | Visit):", p_buy_given_visit)
```

```python
# Independent events

p_a = 0.5
p_b = 0.4

p_a_and_b = p_a * p_b

print("P(A and B):", p_a_and_b)
```

```python
# Dependent events

p_first = 0.5
p_second_given_first = 0.6

p_both = p_first * p_second_given_first

print("P(Both):", p_both)
```

```python
# Bayes theorem

p_disease = 0.01
p_positive_given_disease = 0.90
p_positive = 0.05

p_disease_given_positive = (
    p_positive_given_disease * p_disease
) / p_positive

print("P(Disease | Positive):", p_disease_given_positive)
```

```python
# Random variable

number_of_orders = 5

print("Number of Orders:", number_of_orders)
```

```python
# Discrete variable

number_of_customers = 50

print("Customers:", number_of_customers)
```

```python
# Continuous variable

customer_waiting_time = 4.5

print("Waiting Time:", customer_waiting_time)
```

```python
# Expected value of a fair dice

values = [1, 2, 3, 4, 5, 6]
probabilities = [1/6] * 6

expected_value = sum(
    x * p for x, p in zip(values, probabilities)
)

print("Expected Value:", expected_value)
```

```python
import numpy as np

values = [1, 2, 3, 4, 5, 6]

print("Mean:", np.mean(values))
print("Variance:", np.var(values))
```

```python
## 12. Bernoulli Distribution
from scipy.stats import bernoulli

# Probability of success = 0.7

p = 0.7

print("P(Success):", bernoulli.pmf(1, p))
print("P(Failure):", bernoulli.pmf(0, p))
```

```python
## 13. Binomial Distribution
from scipy.stats import binom

# 10 trials, probability of success = 0.5

probability = binom.pmf(5, 10, 0.5)

print("P(Exactly 5 successes):", probability)
```

```python

## 14. Poisson Distribution
# Poisson distribution

import numpy as np

average_calls = 5

calls = np.random.poisson(average_calls)

print("Calls:", calls)
```

```python
## 15. Normal Distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Create normal distribution

x = np.linspace(-4, 4, 100)
y = norm.pdf(x, 0, 1)

plt.plot(x, y)
plt.title("Standard Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
# Exponential distribution
import numpy as np
average_wait = 5
wait_time = np.random.exponential(average_wait)
print("Wait Time:", wait_time)
```

```python
# Uniform distribution
import numpy as np
p_min = 1
p_max = 5

rating = np.random.uniform(p_min, p_max)

print("Rating:", rating)
```
